```python
username=input()
password=input()
while True:
  u=input("enter username :")
  p=input("enter password :")
  if password==p and username==u:
   print("login succesfull")
   break
  elif username==u and password!=p:
   print("invalid pass")
  elif username!=u and password==p:
   print("invalid username")
  elif username!=u and password!=p:
   print("invalid Details")
  else:
   print("Enter correct Details")
```

```python
# guessing number program
number=int(input())
while True:
  gess=int(input("enter gess number"))
  if gess > 10:
    print("invalid gess")
    break
  if number==gess:
    print("gessing number unlock")
    break
  elif gess>number:
    print("your number is greater than my number")
  elif gess<number:
    print("your number is smaller than my number")
  else:
    print("not match try number")
```

```python
#atm pin verification
atm=int(input())
while True:
  if atm<999 or atm>9999:
    print("invalid pin")
    break
  pin=int(input("enter your pin :"))
  if atm==pin:
    print("money credit")
    break
  else:
    print("incorrect pin")
```
