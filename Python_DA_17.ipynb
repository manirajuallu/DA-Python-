Dictionary in python
A Dictionary is a bulit-in data structure in python used to store data in the from key-value pairs
1.Each value is associated with a unique key
2.Dictionaries are mutable(we can modify them)
3.Dictiomaries are ordered(maintain insertion order)
4.Keys must be unique
5.Values can be duplicated
6.Keys Should be immutable data type such as

```python
#Dictionary
std={"name":"rahul","age":22,"marks":75}
print(std)
print(std["name"])
print(std["age"])
print(std["marks"])
print(type(std))
```

```python
#Adding New element
std["course"]="pyhton"
print(std)
#updating the elements
std["marks"]=100
print(std)
```

```python
#comverting the list into dict
l=["mani",75,98]
print(l)
dic={i: value for i, value in enumerate(l)}
print(dic)
```

```python
#creating a dictionary of employee details with 5 details?
emp={"name":"mani","age":25,"salary":100000,"company":"mnc","emp_id":101}
print(emp)
#updating the values
emp["salary"]=120000
print(emp)
#adding the new value
emp["blood_Grp"]="B+"
print(emp)
```

```python
#removing the value
emp.pop("age")
print(emp)
emp.pop("salary")
print(emp)
```

```python
emp.popitem()
print(emp)
```

```python
emp.clear()
print(emp)
```

```python
for i in std.items():
  print(i)
```

```python
for i,j in std.items():
  print(f'{i} :{j}')
```

```python
print(std.get('name'))
print(std.get('age'))
```
