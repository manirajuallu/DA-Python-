```python
#Single Inheritance:
class Employe:
  def display(self,name,salary):
    self.name=name
    self.salary=salary
    print('employee details :',name,salary)
class DataAnalyst(Employe):
  def analysis(self,course):
    self.course=course
    print('Analyzing Data:',course)
obj=DataAnalyst()
obj.display("mani",90876666)
obj.analysis("python")
```

```python

## Create a mentor class with subject,name,salary and child class as student course,marks ?
#Multi level
class mentor:
  def sir(self,subject,name,salary):
    self.name=name
    self.salary=salary
    print('Sub detials:',subject,name,salary)
class student(mentor):
  def std(self,course,marks):
    self.course=course
    self.marks=marks
    print("Std Details :",course,marks)
class mani(student):
  def greet(self,laptop):
    self.laptop=laptop
    print('Greet Details :',laptop)
obj=mani()
obj.sir('Python','Sai',27686676567)
obj.std('DA',87)
obj.greet('ACER')


```

```python
#Create a Class Person with a method Details().Create a class Employee that inherits from person and contains a method work().Create a class Manager that inherits from employee and contain and manage().
class Person:
  def Details(self,name,age,number):
    print('Person Details :',name,age,number)
class Employee(Person):
  def work(self,job):
    print('Work Details :',job)
class Manager(Employee):
  def manage(self,Role):
    print('Manage Role :',Role)
obj=Manager()
obj.Details("Mani",20,9876543210)
obj.work('IT')
obj.manage('Data Analytics')
```
