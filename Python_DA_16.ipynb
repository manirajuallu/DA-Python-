```python
#Set
s={1,2,3,4,5,6,2}
print(s)
```

```python
l=[1,2,3,4,56,6,1,3,4,5]
print(l)
t=set(l)
print(type(t))
print(t)
```

```python
t=(1,2,4,56,7,56)
print(type(t))
print(t)
s=set(t)
print(type(s))
print(s)
```

```python
a=set([10,20,30])
print(a)
print(type(a))
```

```python
a={10,20.5,"python",True}
print(a)
```

```python
#Accsing elements in set
a={1,2,3,4,5,6,7,8}
for i in a:
  print(i)
```

```python
#add one elements into set
s={10,20,30,40}
print(s)
s.add(50)
print(s)
```

```python
#updating the set
s.update([70,80])
print(s)
s.update([90])
print(s)
```

```python
#removing
s.remove(80)
print(s)
```

```python
#discard
f={10,20,30,40,50}
f.discard(50)
print(f)
```

```python
#POP is used delete random elements
f.pop()
print(f)
```

```python
#clear are use to remove all elements
f.clear()
print(f)
```

```python
#copy means create a shallow copy of the list
s={10,20,30,40,50}
b=s.copy()
print(s)
print(b)
```

```python
#Mathematical operations on set
#Union(|)
a={1,2,4,3,5,6,7}
b={1,2,5,8,9,45,67}
print(a|b)
print(a.union(b))
```

```python
#Difference on set(-)
print(a.difference(b))
print(a-b)
```

```python
#intersection(&)
print(a&b)
```

```python
#symmetric(^)
print(a^b)
print(a.symmetric_difference(b))
```
