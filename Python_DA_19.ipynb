```python
#LAMBDA FUNCTION
x=lambda a:a*a
print(x(4))
square=lambda a,b:a+b
print(square(4,8))
```

```python
x=lambda a,b,c:a*b*c
print(x(2,3,4))
x=lambda a,b,c:a-b-c
print(x(2,3,4))
x=lambda a,b,c:a+b+c
print(x(2,3,4))
```

```python
x=lambda a,b:a-b
print(x(4,3))
x=lambda a,b:a+b
print(x(4,3))
x=lambda a,b:a%b
print(x(4,4))
```

```python
check =lambda a:"even" if a%2==0 else'odd'
print(check(4))
print(check(9))
```

```python
#Given a list of number,use map() and lambda to add 10 at every element, then use filter() to select number given greater than 20
s=[10,20,40,50]
l=list(map(lambda a:a+10,s))
print(l)
r=list(filter(lambda a:a>20,l))
print(r)
```
